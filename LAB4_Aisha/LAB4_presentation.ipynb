## Task 1 - Driver and Team classes

```python
# make a driver and print it
# should print Carlos Sainz (200 pts)
sainz = Driver("Carlos Sainz", 200)
print(sainz.name)
print(sainz.points)
print(sainz)
```

```python
# make a team, it should start with no drivers
ferrari = Team("FERRARI")
print(ferrari.drivers)
```

```python
# add 2 drivers. total should be 200 + 206 = 406
leclerc = Driver("Charles Leclerc", 206)
ferrari.add_driver(sainz)
ferrari.add_driver(leclerc)

print(ferrari.drivers)
print(ferrari.get_total_points())
print(ferrari)
```

```python
# __lt__ test. haas has 12 points and ferrari has 406
# so haas < ferrari should be True and ferrari < haas should be False
haas = Team("HAAS FERRARI")
haas.add_driver(Driver("Kevin Magnussen", 3))
haas.add_driver(Driver("Nico Hulkenberg", 9))

print(haas < ferrari)
print(ferrari < haas)
```

## Task 2 - load the file

This is what the start of the file looks like.

```python
file = open('f1_points.csv', 'r')
text = file.read()
file.close()

lines = text.split('\n')
for line in lines[:4]:
    print(line)
```

```python
# how one line gets split up into the 3 parts
line = lines[1]
driver_name, team_name, points = line.split(',')
points = int(points)

print(driver_name)
print(team_name)
print(points)
```

load_teams does that for every line. The teams go in a dictionary where the key is the team name. It checks if the team name is already in the dictionary before making a new Team, so each team only gets made once.

```python
# should be 10 teams
teams = load_teams('f1_points.csv')
print(len(teams))
```

```python
# every team should have 2 or more drivers
for team_name in teams:
    print(team_name, len(teams[team_name].drivers))
```

```python
# check one team with the file. ferrari has Sainz (200) and Leclerc (206) so 406
print(teams['FERRARI'])
print(teams['FERRARI'].drivers)
```

```python
# the file has 22 drivers, so all the teams together should have 22
total_drivers = 0
for team in teams.values():
    total_drivers += len(team.drivers)

print(total_drivers)
```

## Task 3 - sort the teams

sorted uses the __lt__ method to compare the teams, so the team with the least points comes first.

```python
sorted_teams = sorted(teams.values())

for team in sorted_teams:
    print(team)
```

```python
# most points first
for team in sorted(teams.values(), reverse=True):
    print(team.name, team.get_total_points())
```

## More tests

Normal case: 3 small teams with 30, 10 and 20 points. Should come out in the order B, C, A.

```python
team_a = Team("A")
team_a.add_driver(Driver("a1", 10))
team_a.add_driver(Driver("a2", 20))

team_b = Team("B")
team_b.add_driver(Driver("b1", 4))
team_b.add_driver(Driver("b2", 6))

team_c = Team("C")
team_c.add_driver(Driver("c1", 20))
team_c.add_driver(Driver("c2", 0))

for team in sorted([team_a, team_b, team_c]):
    print(team)
```

Special cases

```python
# a team with no drivers should have 0 points
empty_team = Team("NEW TEAM")
print(empty_team.get_total_points())
print(empty_team < team_b)
```

```python
# a driver with 0 points (Nyck De Vries) should still be on his team
# alphatauri is also the team with 4 drivers. total should be 6 + 2 + 0 + 17 = 25
alphatauri = teams['ALPHATAURI HONDA RBPT']
print(alphatauri.drivers)
print(len(alphatauri.drivers))
print(alphatauri.get_total_points())
```

```python
# two teams with the same points. neither one is less than the other so both should be False
tie_1 = Team("TIE 1")
tie_1.add_driver(Driver("t1", 10))
tie_2 = Team("TIE 2")
tie_2.add_driver(Driver("t2", 10))

print(tie_1 < tie_2)
print(tie_2 < tie_1)
```

Error cases

```python
# adding a string instead of a Driver object should raise a TypeError
test_team = Team("TEST")
try:
    test_team.add_driver("Max Verstappen")
except TypeError as error:
    print("TypeError:", error)

# and nothing should have been added
print(test_team.drivers)
```
